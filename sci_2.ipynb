```python
import numpy as np
from scipy import linalg
```

7x + 2y = 8 and 
4x + 5y = 10

```python
a=np.array([[7,2],[4,5]])
b=np.array([8,10])


x=linalg.solve(a,b)
print(x)
```

```python
#inverse of a matrix

a=[[3,6],[5,10]]
print(a)

print("inverse")
print(linalg.inv(a))

```

```python
a=np.array([[9,6],[4,5]])
det_a=linalg.det(a)
print(det_a)
```

```python
M = np.array([[9, 3], [2, 4]])
eigenvalues, eigenvectors = linalg.eig(M)

print("Eigenvalues:", eigenvalues)
print("Eigenvectors:\n", eigenvectors)
```
