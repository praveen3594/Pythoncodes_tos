```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x=np.array([3,5,7,4])
y=np.array([5,7,1,7])
```

```python
plt.scatter(x,y, color="red")
plt.show()
```

```python
plt.bar(x,y)
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.show()
```

```python
x = ["APPLES", "BANANAS"]
y = [400, 350]
plt.bar(x, y)

```

```python
x = np.array(["A", "B", "C", "D"])
y = np.array([3, 8, 1, 10])

plt.barh(x,y, color="green")
plt.show()
```
