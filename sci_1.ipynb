scipy.status is a module that provides functions for statistical analysis , probability distributions, hypothesis testing , correlation analysis and regression

```python
from scipy import stats
import numpy as np
```

```python
data=np.array([10,12,15,18,20])

stats.tmean(data)
stats.mode(data)
stats.tvar(data) #variance
stats.tstd(data)
stats.skew(data)

stats.tmin(data)
stats.tmax(data)

```

```python
stats.describe(data)
```
