```python
from scipy.optimize import minimize
```

```python
def f(x):
    return x**2 + 4*x + 5

result=minimize(f,x0=0)
print(result.x) #value of x where we get minimum value
print(result.fun) #minimum value of the function

```

```python
def f(v):
    x,y=v
    return x**2+y**2

result=minimize(f,[4,5])


print(result.x)
```

```python
def f(x):
    return (x-5)**2

bounds = [(0,15)]

result = minimize(f, x0=1, bounds=bounds)

print(result.x)
```

```python
def f(v):
    x, y = v
    return x**2 + y**2

constraint = {
    "type":"eq",
    "fun":lambda v: v[0]+v[1]-1
}

result = minimize(f, [0,0], constraints=constraint)
print(result.x)
```

```python
from scipy.optimize import root
def f(x):
    return x**2 - 9

result=root(f,5)
print(result.x)
```

```python
from scipy.optimize import root_scalar

def f(x):
    return  x**2 - 9

result=root_scalar(f,bracket=[0,5])
print(result.root)
```
