```python
import matplotlib.pyplot as plt
import numpy as np

x=np.array([0,6])
y=np.array([0,250])
plt.plot(x,y)
plt.show()
```

```python
plt.plot(x,y,'o')
plt.show()
```

```python
x=np.array([1,2,6,8])
y=np.array([3,8,1,10])

plt.plot(x,y)
plt.show()
```

```python
y=np.array([3,8,1,10,5,7])
plt.plot(y)
plt.show()
```

```python
y=np.array([3,8,1,10])

plt.plot(y,marker='o')
plt.show()
```

```python
y=np.array([3,4,7,9])
plt.plot(y,marker="*")
plt.show()
```

```python
y=np.array([3,5,7,9])
plt.plot(y)
plt.show()
```

```python
y=np.array([3,6,5,7])
plt.plot(y,'o:r')
plt.show()
```

```python
y=np.array([3,5,4,8])
plt.plot(y,marker='o',ms=20,mec='g' ,mfc='r')
plt.show()
```

```python
y= np.array([3, 8, 1, 10])


plt.plot(y, linestyle='dotted' ,c='red', linewidth='1')
plt.show()
```

```python
y1 = np.array([3, 8, 1, 10])
y2 = np.array([6, 2, 7, 11])


plt.plot(y1)
plt.plot(y2)
plt.show()
```

```python
x1 = np.array([0, 1, 2, 3])
y1 = np.array([3, 8, 1, 10])
x2 = np.array([0, 1, 2, 3])
y2 = np.array([6, 2, 7, 11])


plt.plot(x1,y1,x2,y2)
plt.show()
```
